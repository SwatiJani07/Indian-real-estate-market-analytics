```python
df["Parking"] = df["Parking"].fillna("Unknown")
print(df["Parking"])
```

```python
df["Facing"] = df["Facing"].fillna("Unknown")
print(df["Facing"]) 
```

```python
df["Balconies"] = df["Balconies"].fillna(df["Balconies"].median())
print(df["Balconies"]) 
```

```python
df["Furnishing"] = df["Furnishing"].fillna(df["Furnishing"].mode()[0])
print(df["Furnishing"]) 
```

```python
df["Locality"] = df["Locality"].fillna("Unknown")
print(df["Locality"])
```

```python
df["Latitude"] = df["Latitude"].fillna(df.groupby("City")["Latitude"].transform("median"))
print(df["Latitude"])
```

```python
df["Longitude"] = df["Longitude"].fillna(df.groupby("City")["Longitude"].transform("median"))
print(df["Longitude"])
```

```python
print(df.columns.tolist())
```

```python
df["Bathrooms"] = df["Bathrooms"].fillna(df["Bathrooms"].median())
print(df["Bathrooms"])
```

```python
city_count = df["City"].value_counts()

print(city_count)
```

```python
city_price_crore = (
    df.groupby("City")["Price_INR"]
      .mean()
      .div(10000000)
        #divided-by-crore-as-its-in-rs
      .sort_values(ascending=False)
      .round(2)
)

print(city_price_crore)
```

```python
#visulization
#1. city wise analysis
plt.bar(city_price_crore.index, city_price_crore.values)

plt.title("Average Property Price by City")
plt.xlabel("City")
plt.xticks(rotation=45)#maintains space between names ofcity in x axis
plt.ylabel("Average Price (₹ Crore)")

plt.show()
```

```python
#2.property wise analysis
property_price = (
    df.groupby("PropertyType")["Price_INR"]
      .mean()
      .div(10000000)
      .sort_values(ascending=False)
      .round(2)
)

print(property_price)#property type wise prices
```

```python
plt.bar(property_price.index,property_price.values)
plt.title("Property wise analysis")
plt.xlabel("Property")
plt.ylabel("Prices")
plt.xticks(rotation=45)
plt.show()
```

```python
#3.Bhk wise visulization
bhk_price = (
    df.groupby("BHK")["Price_INR"]
      .mean()
      .div(10000000)
      .sort_values(ascending=False)
      .round(2)
)

print(bhk_price)
```

```python
plt.bar(bhk_price.index,bhk_price.values)
plt.title("Bhk wise Analysis")
plt.xlabel("Bhk")
plt.ylabel("prices")
plt.xticks(rotation=45)
plt.show()
```

```python
price_range = pd.cut(
    #cut is used to divide in group and in particular range
    df["Price_INR"],
    bins=[0, 5000000, 10000000, 20000000, 50000000, float("inf")],
    labels=["0-50L", "50L-1Cr", "1-2Cr", "2-5Cr", "5Cr+"]
)

price_count = price_range.value_counts().sort_index()

print(price_count)
```

```python
plt.figure(figsize=(10, 6))

plt.bar(price_count.index, price_count.values)

plt.title("Number of Properties by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Number of Properties")

plt.show()
```

```python
#16.
city_rera = pd.crosstab(df["City"], df["IsRERARegistered"])

print(city_rera)

city_rera.plot(kind="bar", figsize=(12, 7))

plt.title("RERA Registration by City")
plt.xlabel("City")
plt.ylabel("Number of Properties")

plt.show()
```

```python
#17. prices per square ft
df["Price_per_sqft"] = df["Price_INR"] / df["SuperBuiltUpArea_sqft"]

price_per_sqft_city = (
    df.groupby("City")["Price_per_sqft"]
      .mean()
      .round(2)
      .sort_values(ascending=False)
)

print(price_per_sqft_city)

plt.figure(figsize=(10, 6))

plt.bar(price_per_sqft_city.index, price_per_sqft_city.values)

plt.title("Average Price per Sqft by City")
plt.xlabel("City")
plt.ylabel("Average Price per Sqft (₹)")

plt.show()
```
